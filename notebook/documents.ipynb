# LangChain Document Structure

This notebook covers the basics of working with documents in LangChain for RAG applications.

## 1. What is a Document?

In LangChain, a `Document` is a simple object with two attributes:
- `page_content`: The text content (string)
- `metadata`: A dictionary with info about the document (source, page number, etc.)

```python
from langchain_core.documents import Document

# Create a simple document
doc = Document(
    page_content="This is the text content of my document i am using langchain to create RAG",
    metadata={"source": "example.txt", "page": 1, "author": "Reem", "created_at": "2026-01-01"}
)

print("Content:", doc.page_content)
print("Metadata:", doc.metadata)
```

## 2. Document Loaders

LangChain provides many loaders to read documents from different sources.

### 2.1 Loading txt Files

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_data_structures.txt",encoding="utf-8")
doc = loader.load()
print(doc)
  
# docs = loader.load()
# print(f"Loaded {len(docs)} pages")
```

### 2.2 Loading Multiple Files from a Directory

```python
from langchain_community.document_loaders import DirectoryLoader
loader = DirectoryLoader("../data/text_files", glob="*.txt", loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"}, show_progress=False)
doc = loader.load()
doc
```

## 2.3 loading pdf

## 3. Text Splitters

Documents are often too long for LLMs. We split them into smaller chunks.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Create a sample long document
long_text = "This is a very long document. " * 100
doc = Document(page_content=long_text, metadata={"source": "example"})

# Split into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,       # Max characters per chunk
    chunk_overlap=20,     # Overlap between chunks
    length_function=len,
    separator=" ,
)

chunks = text_splitter.split_documents([doc])
print(f"Split into {len(chunks)} chunks")
print(f"First chunk: {chunks[0].page_content[:100]}...")
```
